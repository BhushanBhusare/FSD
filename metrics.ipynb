```python
import numpy as np

X = [0,1,0,1,0,0,0,1,1,0]
Y = [0,0,1,1,0,1,1,0,1,1]

TP = sum(1 for i in range(len(X)) if X[i]==1 and Y[i]==1)
TN = sum(1 for i in range(len(X)) if X[i]==0 and Y[i]==0)
FP = sum(1 for i in range(len(X)) if X[i]==0 and Y[i]==1)
FN = sum(1 for i in range(len(X)) if X[i]==1 and Y[i]==0)

print("True Positives:", TP)
print("True Negatives:", TN)
print("False Positives:", FP)
print("False Negatives:", FN)

accuracy = (TP + TN) / (TP + TN + FP + FN) * 100
precision = TP / (TP + FP) * 100
recall = TP / (TP + FN) * 100

F1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"Accuracy: {accuracy:.2f}%")
print(f"Precision: {precision:.2f}%")
print(f"Recall: {recall:.2f}%")
print(f"F1 Score: {F1_score:.2f}%")

```
