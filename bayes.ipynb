```python
from sklearn.preprocessing import LabelEncoder

from sklearn.naive_bayes import MultinomialNB

X = [[2, 50], [8, 85], [3, 60], [7, 90], [1, 45], [9, 95], [4, 65], [6, 80]]

Y = ["Fail", "Pass", "Fail", "Pass", "Fail", "Pass", "Fail", "Pass"]

le = LabelEncoder()
Y = le.fit_transform(Y)

model = MultinomialNB()
model.fit(X,Y)

n_s = [[2,50]]
proba = model.predict_proba(n_s)
predict = model.predict(n_s)

print(f"Classes: {le.classes_}")
print(f"P_fail: {proba[0][0]}")
print(f"P_pass: {proba[0][1]}")
print(f"Predict: {le.inverse_transform(predict)[0]}")
```
