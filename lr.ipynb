```python
x = [5, 7, 12, 16, 20]
y = [40, 120, 180, 210, 240]

#mean
n = len(x)
x_mean = sum(x)/n
y_mean = sum(y)/n

#deviation
a = [i - x_mean for i in x]
b = [i - y_mean for i in y]

# Sxy and Sxx
Sxy = sum(a[i] * b[i] for i in range(n))
Sxx = sum(i**2 for i in a)

B = Sxy/Sxx
A = y_mean - (B * x_mean)

print(f"Regression: {round(A,2)}+{round(B,2)}x")
x_in = int(input("Enter X: "))
print(f"Y: {A+(B*x_in)}")
```
